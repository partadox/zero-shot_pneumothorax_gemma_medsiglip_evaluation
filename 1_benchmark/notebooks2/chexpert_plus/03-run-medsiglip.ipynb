# 03 — CheXpert Plus: MedSigLIP Inference
Zero-shot PTX classification via cosine similarity (~1GB VRAM).
Resume-safe: existing results in JSONL are skipped automatically.

```python
import os, sys
from pathlib import Path

PROJECT_ROOT = Path(os.getcwd()).parent.parent
sys.path.insert(0, str(PROJECT_ROOT))

from src2.utils import load_config, load_done_ids
from src2.data import load_chexpert_plus_benchmark
from src2.medsiglip_runner import MedSigLIPRunner

CFG         = load_config(PROJECT_ROOT / 'configs2' / 'server.yaml')
DATA_ROOT   = Path(CFG['paths']['data_root'])
OUTPUT_ROOT = Path(CFG['paths']['output_root'])
OUT_JSONL   = OUTPUT_ROOT / CFG['outputs']['chexpert_plus']['medsiglip']
OUT_JSONL.parent.mkdir(parents=True, exist_ok=True)

print('DATA_ROOT  :', DATA_ROOT)
print('Output JSONL:', OUT_JSONL)
```

```python
df = load_chexpert_plus_benchmark(DATA_ROOT, seed=42)

done_ids = load_done_ids(OUT_JSONL)
print(f'Total: {len(df)} | Already done: {len(done_ids)} | Remaining: {len(df) - len(done_ids)}')
```

```python
runner = MedSigLIPRunner(model_path=CFG['models']['medsiglip'])
```

```python
runner.run_dataset(df, OUT_JSONL, dataset_name='chexpert_plus', done_ids=done_ids)
```

```python
runner.free()
print('Done. JSONL:', OUT_JSONL)
```
